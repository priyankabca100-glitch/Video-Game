```python


# VIDEO GAME SALES PREDICTION USING MACHINE LEARNING


# 1. IMPORT LIBRARIES

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display, Markdown

from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


# 2. LOAD DATASET

df = pd.read_csv("vgsales.csv")

print("DATASET LOADED SUCCESSFULLY")
print("Original Dataset Shape:", df.shape)

display(df.head())


# 3. UNDERSTANDING THE DATASET

print("\nDATASET INFORMATION")
df.info()

print("\nDATASET DESCRIPTION")
display(df.describe(include="all"))

print("\nMISSING VALUES")
display(df.isnull().sum())

print("\nDUPLICATE ROWS:", df.duplicated().sum())

print("\nCOLUMN NAMES")
print(df.columns.tolist())


# 4. DATA CLEANING

# Remove duplicate records
df = df.drop_duplicates()

# Convert required columns into correct data types
df["Global_Sales"] = pd.to_numeric(
    df["Global_Sales"], errors="coerce"
)

df["Year"] = pd.to_numeric(
    df["Year"], errors="coerce"
)

# Remove rows where target value is missing
df = df.dropna(subset=["Global_Sales"])

# Remove invalid negative sales values
df = df[df["Global_Sales"] >= 0]

# Keep only rows with valid target values
df = df.reset_index(drop=True)

print("\nCLEANED DATASET")
print("Shape after cleaning:", df.shape)

print("\nMissing values after cleaning:")
display(df.isnull().sum())


# 5. EXPLORATORY DATA ANALYSIS (EDA)

# Graph 1: Distribution of Global Sales

plt.figure(figsize=(9, 5))
sns.histplot(df["Global_Sales"], bins=50, kde=True)
plt.title("Distribution of Global Video Game Sales")
plt.xlabel("Global Sales")
plt.ylabel("Number of Games")
plt.show()

print("Observation: This graph shows the distribution of global sales.")


# Graph 2: Sales by Genre

genre_sales = (
    df.groupby("Genre")["Global_Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(11, 5))
genre_sales.plot(kind="bar")
plt.title("Total Global Sales by Genre")
plt.xlabel("Genre")
plt.ylabel("Global Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Highest selling genre:", genre_sales.idxmax())


# Graph 3: Sales by Platform

platform_sales = (
    df.groupby("Platform")["Global_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))
platform_sales.plot(kind="bar")
plt.title("Top 10 Platforms by Global Sales")
plt.xlabel("Platform")
plt.ylabel("Global Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Highest selling platform:", platform_sales.idxmax())


# Graph 4: Sales by Release Year

year_sales = (
    df.dropna(subset=["Year"])
    .groupby("Year")["Global_Sales"]
    .sum()
    .sort_index()
)

plt.figure(figsize=(12, 5))
plt.plot(year_sales.index, year_sales.values, marker="o")
plt.title("Global Video Game Sales by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Global Sales")
plt.grid(True)
plt.show()

print("Observation: This graph shows how total sales vary across release years.")


# 6. FEATURE SELECTION

# Input features
X = df[["Platform", "Genre", "Year", "Publisher"]]

# Target variable
y = df["Global_Sales"]

print("\nINPUT FEATURES:")
print(X.head())

print("\nTARGET VARIABLE:")
print(y.head())


# 7. TRAIN-TEST SPLIT (80:20)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTRAINING DATA:", X_train.shape)
print("TESTING DATA:", X_test.shape)


# 8. DATA PREPROCESSING

categorical_features = ["Platform", "Genre", "Publisher"]
numerical_features = ["Year"]

# Handle missing numerical and categorical values
# Convert categorical values into numerical form

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)


# 9. CREATE THREE REGRESSION MODELS

models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )
}

trained_models = {}
results = []


# 10. TRAIN AND EVALUATE THREE MODELS

for name, model in models.items():

    pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", model)
    ])

    # Train model
    pipeline.fit(X_train, y_train)

    # Predict test data
    predictions = pipeline.predict(X_test)

    # Calculate evaluation metrics
    mae = mean_absolute_error(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, predictions)

    trained_models[name] = pipeline

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

    print("\nMODEL:", name)
    print("MAE:", round(mae, 4))
    print("MSE:", round(mse, 4))
    print("RMSE:", round(rmse, 4))
    print("R2 Score:", round(r2, 4))


# 11. MODEL COMPARISON TABLE

comparison = pd.DataFrame(results)

print("\nMODEL COMPARISON")
display(comparison.sort_values(by="MAE"))


# 12. FIVE-FOLD CROSS VALIDATION

print("\nFIVE-FOLD CROSS VALIDATION RESULTS")

cv_results = []

for name, pipeline in trained_models.items():

    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        cv=5,
        scoring="neg_mean_absolute_error",
        n_jobs=-1
    )

    mean_mae = -scores.mean()

    cv_results.append({
        "Model": name,
        "Mean CV MAE": mean_mae
    })

    print(name, "Mean CV MAE:", round(mean_mae, 4))

cv_table = pd.DataFrame(cv_results)
display(cv_table)


# 13. HYPERPARAMETER TUNING - DECISION TREE

print("\nDECISION TREE HYPERPARAMETER TUNING")

dt_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", DecisionTreeRegressor(random_state=42))
])

dt_parameters = {
    "model__max_depth": [10, 20, None],
    "model__min_samples_split": [2, 5]
}

dt_grid = GridSearchCV(
    dt_pipeline,
    dt_parameters,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)

dt_grid.fit(X_train, y_train)

best_dt = dt_grid.best_estimator_

print("Best Decision Tree Parameters:")
print(dt_grid.best_params_)

print("Best Decision Tree CV MAE:")
print(round(-dt_grid.best_score_, 4))


# 14. HYPERPARAMETER TUNING - RANDOM FOREST

print("\nRANDOM FOREST HYPERPARAMETER TUNING")

rf_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        random_state=42,
        n_jobs=-1
    ))
])

rf_parameters = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [10, None]
}

rf_grid = GridSearchCV(
    rf_pipeline,
    rf_parameters,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=1
)

rf_grid.fit(X_train, y_train)

best_rf = rf_grid.best_estimator_

print("Best Random Forest Parameters:")
print(rf_grid.best_params_)

print("Best Random Forest CV MAE:")
print(round(-rf_grid.best_score_, 4))


# 15. EVALUATE TUNED MODELS

tuned_models = {
    "Tuned Decision Tree": best_dt,
    "Tuned Random Forest": best_rf
}

for name, pipeline in tuned_models.items():

    predictions = pipeline.predict(X_test)

    mae = mean_absolute_error(y_test, predictions)
    mse = mean_squared_error(y_test, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })


# Updated comparison including tuned models

final_comparison = pd.DataFrame(results)

print("\nFINAL MODEL COMPARISON")
display(final_comparison.sort_values(by="MAE"))


# 16. CONFUSION MATRIX FOR ALL MODELS

# Sales below 1.0 = Low Sales
# Sales 1.0 or above = High Sales
# Threshold is illustrative.

def sales_category(values):
    return np.where(values < 1.0, "Low Sales", "High Sales")


all_models = {
    **trained_models,
    **tuned_models
}

confusion_results = []

for name, pipeline in all_models.items():

    predictions = pipeline.predict(X_test)

    actual_categories = sales_category(y_test)
    predicted_categories = sales_category(predictions)

    cm = confusion_matrix(
        actual_categories,
        predicted_categories,
        labels=["Low Sales", "High Sales"]
    )

    print("\nCONFUSION MATRIX:", name)

    display_labels = ["Low Sales", "High Sales"]

    display_cm = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=display_labels
    )

    display_cm.plot(cmap="Blues")
    plt.title("Confusion Matrix - " + name)
    plt.show()

    accuracy = accuracy_score(
        actual_categories, predicted_categories
    )

    precision = precision_score(
        actual_categories,
        predicted_categories,
        pos_label="High Sales",
        zero_division=0
    )

    recall = recall_score(
        actual_categories,
        predicted_categories,
        pos_label="High Sales",
        zero_division=0
    )

    f1 = f1_score(
        actual_categories,
        predicted_categories,
        pos_label="High Sales",
        zero_division=0
    )

    confusion_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    })


# Display confusion matrix evaluation metrics

print("\nCONFUSION MATRIX METRICS")
confusion_table = pd.DataFrame(confusion_results)
display(confusion_table)


# 17. SELECT THE BEST MODEL

best_result = final_comparison.loc[
    final_comparison["MAE"].idxmin()
]

best_model_name = best_result["Model"]

print("\nBEST MODEL:", best_model_name)
print("Lowest MAE:", round(best_result["MAE"], 4))
print("RMSE:", round(best_result["RMSE"], 4))
print("R2 Score:", round(best_result["R2 Score"], 4))

if best_model_name in all_models:
    final_model = all_models[best_model_name]


# 18. SAMPLE PREDICTIONS

sample_data = X_test.head(10)

sample_predictions = final_model.predict(sample_data)

prediction_table = pd.DataFrame({
    "Platform": sample_data["Platform"].values,
    "Genre": sample_data["Genre"].values,
    "Year": sample_data["Year"].values,
    "Publisher": sample_data["Publisher"].values,
    "Actual Sales": y_test.head(10).values,
    "Predicted Sales": sample_predictions
})

print("\nSAMPLE PREDICTIONS")
display(prediction_table)


# 19. FINAL PROJECT SUMMARY

print("\n" + "=" * 55)
print("VIDEO GAME SALES PREDICTION - PROJECT SUMMARY")
print("=" * 55)

print("Total Records Used:", len(df))
print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))

print("\nInput Features:")
print("Platform, Genre, Year, Publisher")

print("\nTarget Variable:")
print("Global_Sales")

print("\nModels Trained:")
for model_name in all_models.keys():
    print("-", model_name)

print("\nSelected Best Model:", best_model_name)

print("\nFinal Model Performance:")
print("MAE:", round(best_result["MAE"], 4))
print("MSE:", round(best_result["MSE"], 4))
print("RMSE:", round(best_result["RMSE"], 4))
print("R2 Score:", round(best_result["R2 Score"], 4))

print("\nsuccessfull!")

```
